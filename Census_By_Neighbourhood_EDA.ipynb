## Question 1: which neighbourhood has the highest and lowest population

```python
max_pop_neigh = df.loc[df['Total Population'].idxmax()]
min_pop_neigh = df.loc[df['Total Population'].idxmin()]
print(f"\nNeighbourhood with the largest Population: {max_pop_neigh['NC_Name']} ({max_pop_neigh['Total Population']})")
print(f"Neighbourhood with the smallest Population: {min_pop_neigh['NC_Name']} ({min_pop_neigh['Total Population']})")
```

# Question 2: Racial composition distribution

```python
df.head(10)
```

```python
racial_cols = ['White_pop', 'Black_pop', 'American_pop',
               'Asian_pop', 'Hawn_pop', 'Other_pop', 'Multi_pop']
racial_composition = df[racial_cols].sum()
racial_composition_percent = (racial_composition / df['Total Population'].sum()) * 100

print("\nRacial composition percentages:")
print(racial_composition_percent)
```

```python
# Plot racial composition
plt.figure(figsize=(10, 6))
racial_composition_percent.plot(kind='bar', color='skyblue')
plt.title("Racial Composition Distribution (%)")
plt.ylabel("Percentage")
plt.xlabel("Racial Categories")
plt.xticks(rotation=45)
plt.grid(axis='y')
plt.show()
```

# Question 3: Percentage of population living in poverty

```python
poverty_percent = (df['In_Poverty'].sum() / df['Total Population'].sum()) * 100
print(f"\nPercentage of total population living in poverty: {poverty_percent:.2f}%")
```

# Question 4: Comparison of homeownership and renting

```python
ownership_renting = df[['Owner_pop', 'Renter_pop']].sum()
plt.figure(figsize=(8, 5))
ownership_renting.plot(kind='bar', color=['green', 'orange'])
plt.title("Comparison of Homeownership and Renting")
plt.ylabel("Population")
plt.xlabel("Categories")
plt.grid(axis='y')
plt.show()
```

# Question 5: Correlation between population groups

```python
correlation_matrix = df[racial_cols + ['In_Poverty', 'Owner_pop', 'Renter_pop']].corr()
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix")
plt.show()
```

```python
# Additional Insights:
# Top 5 neighborhoods by total population
top_5_pop = df.nlargest(5, 'Total Population')[['NC_Name', 'Total Population']]
print("\nTop 5 neighborhoods by total population:")
print(top_5_pop)
```

```python
# Step 7: Visualize top 5 neighborhoods
plt.figure(figsize=(10, 8))
sns.barplot(data=top_5_pop, x='NC_Name', y='Total Population', palette='Blues_d')
plt.title("Top 5 Neighborhoods by Total Population")
plt.ylabel("Total Population")
plt.xlabel("Neighborhood")
plt.xticks(rotation=90)
plt.show()
```
